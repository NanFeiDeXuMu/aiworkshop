### Instructor fallback / `src/api.js` review reference

Use this only after students review the generated diff:

```javascript
export const API = import.meta.env.VITE_API_URL || "http://localhost:8000"
export const CHAT_ID = "day2-demo"

async function readJSON(response) {
  const data = await response.json().catch(() => ({}))
  if (!response.ok) throw new Error(data.detail || `Request failed (${response.status})`)
  return data
}

export async function uploadPDF(file) {
  const formData = new FormData()
  formData.append("file", file)
  const response = await fetch(
    `${API}/upload?chat_id=${encodeURIComponent(CHAT_ID)}`,
    { method: "POST", body: formData },
  )
  return readJSON(response)
}

export async function askQuestion(message) {
  const response = await fetch(`${API}/chat`, {
    method: "POST",
    headers: { "Content-Type": "application/json" },
    body: JSON.stringify({ message, chat_id: CHAT_ID }),
  })
  return readJSON(response)
}
```

# Appendix A: React Vocabulary in This Project

## Component

A component is a function that returns UI. `PdfUploader`, `ChatPanel`, and `App` are components. A component name starts with a capital letter so React can distinguish it from an HTML element.

## Prop

A prop is input from a parent. `onUploaded` gives the child a way to report success; `document` gives the chat panel the selected upload metadata. Props flow down.

## State

State is component memory that can trigger a re-render. A normal variable does not tell React to update the page. State should contain information needed across renders, not every temporary value.

## Event handler

An event handler runs because the user acts: selecting a file, submitting a form, or typing. It may update state or start a request.

## Conditional rendering

The expression `document ? <ChatPanel /> : <p>...</p>` makes the product sequence visible: chat exists only after successful upload.

### Label the flow

Take one upload attempt and label each moment as prop, state, event, request, response, setter, or render.


# Appendix B: Controlled Inputs and Forms

A controlled textarea receives its value from React state:

```jsx
<textarea
  value={message}
  onChange={(event) => setMessage(event.target.value)}
/>
```

The state is the source of truth. This lets the app validate, preserve, clear, or resubmit the message intentionally.

### Prediction lab

Predict before testing:

- Remove `onChange` but keep `value`.
- Remove `value` but keep `onChange`.
- Put `setMessage("")` before the network request.
- Call `event.preventDefault()` after asynchronous work instead of immediately.

Observe one change at a time. Restore the working version after each experiment.

### Design decision

In this workshop, preserve the message when a request fails. Clearing it would make recovery harder. Clearing after success is optional; choose intentionally and state the behavior in the acceptance criteria.


# Appendix C: FormData versus JSON with `fetch`

The app uses two body formats for two jobs.

## Upload

```javascript
const formData = new FormData()
formData.append("file", file)
await fetch(`${API}/upload?chat_id=${encodeURIComponent(CHAT_ID)}`, {
  method: "POST",
  body: formData,
})
```

The browser creates the multipart boundary automatically.

## Chat

```javascript
await fetch(`${API}/chat`, {
  method: "POST",
  headers: { "Content-Type": "application/json" },
  body: JSON.stringify({ message: message.trim(), chat_id: CHAT_ID }),
})
```

The JSON field names must match the backend contract exactly.


# Appendix D: Stretch Component Review

The single `App.jsx` implementation is the Day 2 core. The optional component below is only a refactoring reference after the core flow passes; equivalent structures are acceptable if they preserve `src/api.js`, fixed `CHAT_ID`, and all observable states.


```jsx
// Optional Stretch example — not required for Day 2 core
import { useState } from "react"
import { askQuestion } from "../api"

export default function ChatPanel() {
  const [message, setMessage] = useState("")
  const [answer, setAnswer] = useState(null)
  const [status, setStatus] = useState("idle")
  const [error, setError] = useState("")

  async function handleAsk(event) {
    event.preventDefault()
    if (!message.trim()) return setError("Write a message first.")
    try {
      setStatus("asking")
      setError("")
      setAnswer(await askQuestion(message.trim()))
    } catch (requestError) {
      setError(requestError.message || "Chat failed.")
    } finally {
      setStatus("idle")
    }
  }

  return (
    <form onSubmit={handleAsk}>
      <label htmlFor="message">Message</label>
      <textarea id="message" value={message} onChange={(event) => setMessage(event.target.value)} />
      <button disabled={!message.trim() || status !== "idle"}>
        {status === "asking" ? "Asking…" : "Ask"}
      </button>
      {error && <p role="alert">{error}</p>}
      {answer && <p>{answer.answer}</p>}
      <div>{answer?.citations?.map((page) => <span key={page}>Page {page}</span>)}</div>
    </form>
  )
}
```


# Appendix E: Accessibility and Honest UI States

Even a workshop prototype should communicate clearly:

- connect every input with a visible `label`;
- use real `button`, `form`, `h1`, and `h2` elements;
- keep keyboard focus visible;
- place errors in an element with `role="alert"`;
- do not use color as the only success/error signal;
- preserve readable contrast;
- use `disabled` during an active request;
- write loading text that describes the operation (“Uploading…”, “Reading…”).

### Test without a mouse

Reload the page, press Tab through controls, choose a file, submit, and reach the message area. Can you always see where focus is? Does Enter submit the intended form? Can you understand errors without relying only on red color?

### Honest status text

Do not display “AI is analyzing every detail” when the backend merely extracts text and sends it to a model. Accurate microcopy builds trust and exposes system boundaries.


# Appendix F: Network Debugging Worksheet

For one successful upload and one successful ask, record:

| Field | Upload | Ask |
|---|---|---|
| Method | | |
| Full URL | | |
| Request body type | | |
| Key input fields | | |
| Status | | |
| Response fields | | |
| Duration | | |

Then record one controlled failure. Distinguish:

- **no request** — handler/validation/render problem;
- **request blocked** — origin/network problem;
- **4xx response** — client input or lifecycle problem;
- **5xx response** — backend/upstream problem;
- **200 but no UI update** — state/render problem.

The browser console may show a dramatic message, but the Network response usually contains the most structured evidence.
